```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import sklearn as sl

from sklearn.datasets import make_circles, make_moons
```

## Generate Toy Datasets

```python
# generate moon dataset
X_moon, y_moon = make_moons(500, noise=0.025)

# generate circle dataset 
X_circle, y_circle = make_circles(500, noise=0.025)
```

```python
# plot datasets
fig, ax = plt.subplots(1, 2, figsize=(14, 7))

# plot circle data
ax[0].scatter(X_circle[... , 0], X_circle[... , 1], s=0.5, c=y_circle, cmap="berlin")

# plot moon data
ax[1].scatter(X_moon[... , 0], X_moon[... , 1], s=0.5, c=y_moon, cmap="berlin")

```
